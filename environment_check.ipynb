```python
import subprocess

result = subprocess.run(
    ["docker" , "--version"],
    capture_output = True,
    text = True
)

if result.returncode == 0:
    print("docker is installed.")
    print(result.stdout.strip())

else:
    print("docker command failed.")
    print(result.stderr.strip())
        
```
