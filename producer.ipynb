```python
import json
from kafka import KafkaProducer
```

```python
producer = KafkaProducer(
    bootstrap_servers="localhost:9092",
    value_serializer=lambda message: json.dumps(message).encode("utf-8")
)

print("Kafka producer is created successfully.")
```

```python
events = [
    {"event_id": "E101", "user_id": "U101", "event_type": "login",    "device": "Mobile"},
    {"event_id": "E102", "user_id": "U102", "event_type": "purchase", "device": "Web"},
    {"event_id": "E103", "user_id": "U103", "event_type": "search",   "device": "Mobile"},
    {"event_id": "E104", "user_id": "U104", "event_type": "login",    "device": "Web"},
    {"event_id": "E105", "user_id": "U105", "event_type": "search",   "device": "Mobile"},
    {"event_id": "E106", "user_id": "U106", "event_type": "purchase", "device": "Web"},
    {"event_id": "E107", "user_id": "U107", "event_type": "login",    "device": "Mobile"},
    {"event_id": "E108", "user_id": "U108", "event_type": "search",   "device": "Web"},
    {"event_id": "E109", "user_id": "U109", "event_type": "purchase", "device": "Mobile"},
    {"event_id": "E110", "user_id": "U110", "event_type": "login",    "device": "Web"}
]

invalid_event = {
    "user_id": "U111",
    "event_type": "login",
    "device": "Mobile"
}
events.append(invalid_event)

print("Total messages to send:", len(events))
```

```python
topic_name = "app-events-piyush"

for event in events:
    metadata = producer.send(topic_name, value=event).get(timeout=10)
    event_id = event.get("event_id", "<missing>")
    print(f"Sent event {event_id}, offset {metadata.offset}")

producer.flush()
producer.close()
print("All messages have been sent.")
```
